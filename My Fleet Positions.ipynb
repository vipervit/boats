```python
import folium
from lib.common import *
```

```python
boat1='Kfar Saba'
boat2='Azohen Vey'
```

```python
boats=['Azohen Vey', 'Ladoga', 'Petsamo', 'Kfar Saba']
maps=[]
```

```python
for boat in boats:
    data=get_boat_data(boat)
    lat=data['latitude']
    lon=data['longitude'] 
    center=([lat, lon])
    mymap=folium.Map(location=center, zoom_start=4)
    folium.Marker([lat, lon], popup=boat).add_to(mymap)
    maps.append(mymap)
```
